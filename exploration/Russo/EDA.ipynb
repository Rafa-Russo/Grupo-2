# EDA: hospedagem e criminalidade em São Paulo

Análise exploratória conjunta dos datasets de Airbnb e crimes.

```python
from pathlib import Path

import pandas as pd

PROJECT_ROOT = next(
    path for path in (Path.cwd(), *Path.cwd().parents)
    if (path / "data").is_dir()
)
DATA_DIR = PROJECT_ROOT / "data"
```

```python
# data

## Airbnb 
listings = pd.read_csv(DATA_DIR / "listings.csv")
print("Data Airbnb:")
print(listings.columns)
listings.describe()
```

```python
# data

## Crimes
crimes = pd.read_csv(DATA_DIR / "crimes.csv")

# Consolidar DESC_PERIODO e DESCR_PERIODO em uma única coluna PERIODO
crimes['PERIODO'] = crimes['DESCR_PERIODO'].fillna(crimes['DESC_PERIODO'])
crimes['PERIODO'] = crimes['PERIODO'].str.strip().str.upper()

# Remover colunas desnecessárias
crimes.drop(columns=[
    # Consolidadas/Duplicadas
    'DESC_PERIODO', 'DESCR_PERIODO',
    'CD_IBGE', 'COD IBGE',
    
    # Redundantes
    'DATA_REGISTRO',
    'LOGRADOURO', 'NUMERO_LOGRADOURO',
    
    # Militares (não impactam hospedagem)
    'CMD', 'BTL', 'CIA',
    
    # Metadados de importação
    'source_file', 'source_year', 'source_sheet',
    
    # Duplicatas (sufixo errado)
    'NOME_DELEGACIA_CIRCUNCRIÇÃO', 'NOME_DEPARTAMENTO_CIRCUNCRIÇÃO',
    'NOME_SECCIONAL_CIRCUNCRIÇÃO', 'NOME_MUNICIPIO_CIRCUNCRIÇÃO'
], inplace=True)

print("Data crimes:")
print(f"Shape: {crimes.shape}")
print(f"\nColunas: {list(crimes.columns)}")
print("\nDescribe:")
crimes.describe()
```
